```python
from typing import TypedDict

class Contact(TypedDict):
    name:str
    email:str
    phone:str
    
def send_email(contact:Contact) ->None:
    print(f"Sending email to {contact['name']} at {contact['email']}")

contact_info:Contact={
    'name':"Lilei",
    "email":"lilei@qq.com",
    "phone":"15555123123",
    "aaa":'fdasfasdfs'
}

send_email(contact_info)
```
